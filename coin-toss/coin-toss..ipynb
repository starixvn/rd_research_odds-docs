# Giá trị kỳ vọng

Giá trị kỳ vọng (Expected Value) trong cá cược được tính bằng công thức:

$$ E = \sum_{i} p_i * x $$

Trong đó:

- `p`: xác suất xảy ra biến cố
- `x`: là odds không kèm tiền cược


Ví dụ: Hai cửa Chẵn/Lẻ trong trò chơi xóc đĩa có xác suất xảy ra như nhau là `0.5`, tỉ lệ trả thưởng ở 2 cửa đều là `0.96`. Giá trị kỳ vọng khi đặt `$1` vào cửa Chẵn là bao nhiêu?

` E = 0.5 * $0.96 + (-$1) * 0.5 = $0.48 - $0.5 = - $0.02 `

Có nghĩa là trong mọi trường hợp người chơi thắng hay thua thì cũng sẽ mất `- $0.02` cho mỗi `$1` tiền cược.

# Return to Player (RTP)

Mối liên hệ giữa RTP và Giá trị kỳ vọng được biểu diễn bằng công thức:

$$ RTP = Stake - E $$

# Coin toss
## Bài toán

Tung đồng xu có 2 mặt A/B, với xác suất `p = 0.5`, với `odds = biến số`

Khi đồng xu được tung lên thì rớt xuống vào khu vực bên dưới.

```
...........
.....x.....
...xxxxx...
.....x.....
...........
```

Đồng xu xuất hiện khớp với khu vực được đánh dấu `x` thì được trả thưởng `x1.5` odds, không thì trả thưởng như bình thường.

Với bài toán trên thì nhà cái thắng khi nào?

## Lời giải

Gọi xác suất để vào khu vực dấu `x` là `px`, khu vực còn lại là `pn`. Dễ thấy `px + pn = 1`.

Đồng xu A/B có xác suất như nhau nên chỉ cần tính cho mặt A cũng tương tự cho B.

Xác suất cho mặt A thằng bonus là `0.5 * px` và mặt A thắng bình thường `0.5 * pn` hay `0.5 * (1 - px)`.

Xác suất cho mặt A thua là `0.5`.

Khi đó:

$$ E = (0.5 * px) * (1.5 * odds) + (0.5 * (1 - px)) * odds - 0.5 $$

Nhà cái thắng khi E < 0, để đơn giản bài toán thì giới hạn dãy số

`px` = [0.01, 1.00].

`odds` = [0.01, 1].

````python
x_range = range(100)
y_range = range(100)
win_cases = []

for x in x_range:
    px = (x + 1) / 100

    for y in y_range:
        odds = (y + 1) / 100
        e = (0.5 * px) * (1.5 * odds) + (0.5 * (1 - px)) * odds - 0.5
        rtp = 1 + e

        if round(e, 2) < 0:
            win_cases.append([px, odds, e, rtp])

from IPython.display import HTML, display

display(HTML(
   '''<table>
        <tr>
            <th>Px</th>
            <th>Odds (exclude stake)</th>
            <th>E</th>
            <th>RTP</th>
        </tr>
        <tr>{}</tr>
      </table>'''.format(
       '</tr><tr>'.join(
           '<td>{}</td>'.format('</td><td>'.join("{:.4f}".format(cell) for cell in row)) for row in win_cases)
       )
))
````
