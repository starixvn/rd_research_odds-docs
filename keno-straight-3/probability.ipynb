```python
import random

def has_straight(numbers):
    numbers.sort()

    for i in range(len(numbers) - 2):
        if numbers[i + 2] == numbers[i] + 2:
            return True
        
    return False

count = 0
times = 10_000_000

for i in range(0, times):
    numbers = random.sample(range(1,81),20)

    if (has_straight(numbers)):
        count += 1

print("Probability: ", count / times)
```
