```python
from random import getrandbits, randint

multiplier_ratio = 20
multiplier_value = 2
rounds = 10_000_000

# Multiplier Value = 1 + Fee / Probability of Multiplier
print("Expected fee:", (multiplier_value - 1) * multiplier_ratio)

def get_result():
    return [getrandbits(1), getrandbits(1), getrandbits(1), getrandbits(1)]

def has_multiplier(ratio = 20):
    return randint(1, 100) <= ratio

def is_even(numbers):
    return sum(numbers) % 2 == 0


even_win = 0
four_red_win = 0
three_red_win = 0
two_red_win = 0

for i in range(0, rounds):
    result = get_result()
    even = is_even(result)

    if is_even(result):
        multiplier = multiplier_value if has_multiplier(multiplier_ratio) else 1
        even_win += multiplier * 2

    if sum(result) == 4:
        multiplier = multiplier_value if has_multiplier(multiplier_ratio) else 1
        four_red_win += multiplier * 16

    if sum(result) == 3:
        multiplier = multiplier_value if has_multiplier(multiplier_ratio) else 1
        three_red_win += multiplier * 4

    if sum(result) == 2:
        multiplier = multiplier_value if has_multiplier(multiplier_ratio) else 1
        two_red_win += multiplier * (16 / 6)

even_fee = (even_win - rounds) / rounds * 100
four_red_fee = (four_red_win - rounds) / rounds * 100
three_red_fee = (three_red_win - rounds) / rounds * 100
two_red_fee = (two_red_win - rounds) / rounds * 100

print("Even Fee:", even_fee)
print("Four Red Fee:", four_red_fee)
print("Three Red Fee:", three_red_fee)
print("Two Red Fee:", two_red_fee)
```
